#What are the different states of a Promise, and how do they transition?

-
 States of a JavaScript Promise
A Promise is an object representing the eventual completion (or failure) of an asynchronous operation. It can exist in three states, and here's how they transition:
🔹 1. pending
- Initial state when a Promise is created.
- The operation hasn’t completed yet, so the final value is unknown.
- It can transition to either fulfilled or rejected.
🔹 2. fulfilled
- The operation completed successfully.
- The Promise has a resulting value.
- Transition: pending → fulfilled when resolve(value) is called.
🔹 3. rejected
- The operation failed.
- The Promise has a reason (usually an error) for the failure.
- Transition: pending → rejected when reject(error) is called.


🔄 State Diagram
               +---------+
               | pending |
               +----+----+
                    |
       +------------+------------+
       |                         |
+------+-----+           +-------+------+
| fulfilled  |           |   rejected   |
+------------+           +--------------+


Once a Promise is fulfilled or rejected, it is settled and cannot change state again.


#j How does the JavaScript event loop handle Promises differently from setTimeout?

-

Great question, Mukund — this dives straight into the heart of JavaScript’s concurrency model. Let’s break it down in a way that’s modular and practical:
🔄 Event Loop Basics
The event loop is the mechanism that allows JavaScript to handle asynchronous operations. It processes tasks in different queues, notably:
- Macro-task queue (also called Task queue): Includes things like setTimeout, setInterval, I/O tasks, etc.
- Micro-task queue: Includes Promises (.then, .catch, .finally), MutationObserver, and queueMicrotask.
⚡ Difference in Execution
| Feature | setTimeout | Promise (e.g. .then) |
| Queue type | Macro-task | Micro-task |
| Priority | Lower | Higher |
| Executed after | Entire call stack + micro-tasks | Just the current call stack |
| Delay effect | May introduce real delay | Runs immediately after call stack clears |


So even if setTimeout(fn, 0) and Promise.resolve().then(fn) are both scheduled to run "as soon as possible", the Promise wins because it's in the micro-task queue — which gets drained before any macro-tasks.

💡 Practical Example
console.log("Start");

setTimeout(() => {
  console.log("setTimeout");
}, 0);

Promise.resolve().then(() => {
  console.log("Promise");
});

console.log("End");


Output:
Start
End
Promise
setTimeout


Even though both are "async", the Promise resolves first because it’s handled by the micro-task queue, which runs immediately after the current synchronous code.
